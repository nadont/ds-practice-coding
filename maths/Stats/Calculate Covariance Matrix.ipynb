## Calculate Covariance Matrix

Write a Python function to calculate the covariance matrix for a given set of vectors. The function should take a list of lists, where each inner list represents a feature with its observations, and return a covariance matrix as a list of lists. Additionally, provide test cases to verify the correctness of your implementation.

```python
vectors = [[1, 2, 3], [4, 5, 6]]
```

```python
import numpy as np

np.cov(vectors)
```

```python
element_mean = [sum(i)/len(i) for i in vectors]
element_mean
```

```python
n = len(vectors[0])
```

```python
co_var = sum(
    (vectors[0][i] - element_mean[0]) * (vectors[1][i] - element_mean[1])
    for i in range(n)
) / (n - 1)
co_var
```

```python
var = [
    sum((x - mean) ** 2 for x in vector) / (len(vector) - 1)
    for vector, mean in zip(vectors, element_mean)
]
var
```

```python
cov_m = [[var[0], co_var], [co_var, var[1]]]
cov_m
```

```python
#### new

m = len(vectors)
n = len(vectors[0])

means = [sum(i) / len(i) for i in vectors]

cov_m = [[0.0] * m for _ in range(m)]

for i in range(m):
    for j in range(m):
        cov_m[i][j] = sum((vectors[i][k] - means[i]) * (vectors[j][k] - means[j]) for k in range(n)) / (n - 1)
cov_m
```
