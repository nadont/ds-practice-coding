## Implement Gradient Descent Variants with MSE Loss

In this problem, you need to implement a single function that can perform three variants of gradient descent Stochastic Gradient Descent (SGD), Batch Gradient Descent, and Mini Batch Gradient Descent using Mean Squared Error (MSE) as the loss function. The function will take an additional parameter to specify which variant to use. Note: Do not shuffle the data

```python
import numpy as np
import math 
# Sample data
X = np.array([[1, 1], [2, 1], [3, 1], [4, 1]])
y = np.array([2, 3, 4, 5])

# Parameters
learning_rate = 0.01
n_iterations = 100
batch_size = 2

# Initialize weights
weights = np.zeros(X.shape[1])

def gradient_descent(X, y, weights, learning_rate, n_iterations, batch_size=1, method=['batch', 'stochastic', 'mini_batch']):
	# Your code here
	m = X.shape[0]
	if method == 'batch':
		for i in range(n_iterations):
			y_pred = X @ weights
			error = y_pred - y
			weights -= learning_rate * (2/m) * (X.T @ error)
		print(weights)
		return weights
	elif method == 'stochastic':
		for i in range(n_iterations):
			for j in range(m):
				y_pred = X[j] @ weights
				error = y_pred - y[j]
				weights -= learning_rate * (2 * X[j]) * error
		print(weights)
		return weights
	elif method == 'mini_batch':
		for i in range(n_iterations): # Outer loop: Epochs
			# Inner loop: Iterate over all batches within the epoch
			for j in range(0, m, batch_size):
				
				X_batch = X[j:j+batch_size]
				y_batch = y[j:j+batch_size]
				
				y_pred = X_batch @ weights
				error = y_pred - y_batch
				
				gradient = (2 / batch_size) * (X_batch.T @ error)

				weights -= learning_rate * gradient
		print(weights)
		return weights
```

```python
## solution
def gradient_descent(X, y, weights, learning_rate, n_iterations, batch_size=1, method='batch'):
    m = len(y)
    
    for _ in range(n_iterations):
        if method == 'batch':
            # Calculate the gradient using all data points
            predictions = X.dot(weights)
            errors = predictions - y
            gradient = 2 * X.T.dot(errors) / m
            weights = weights - learning_rate * gradient

        elif method == 'stochastic':
            # Update weights for each data point individually
            for i in range(m):
                prediction = X[i].dot(weights)
                error = prediction - y[i]
                gradient = 2 * X[i].T.dot(error)
                weights = weights - learning_rate * gradient

        
        elif method == 'mini_batch':
            # Update weights using sequential batches of data points without shuffling
            for i in range(0, m, batch_size):
                X_batch = X[i:i+batch_size]
                y_batch = y[i:i+batch_size]
                predictions = X_batch.dot(weights)
                errors = predictions - y_batch
                gradient = 2 * X_batch.T.dot(errors) / batch_size
                weights = weights - learning_rate * gradient
    print(weights)
                
    return weights
```
