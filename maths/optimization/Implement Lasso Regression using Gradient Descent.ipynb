## Implement Lasso Regression using Gradient Descent

Implement the Lasso Regression algorithm using Gradient Descent. Lasso Regression (Least Absolute Shrinkage and Selection Operator) uses L1 regularization, which adds a penalty equal to the absolute value of the coefficients to the loss function. 

```python
import numpy as np

X = np.array([[0, 0], [1, 1], [2, 2]])
y = np.array([0, 1, 2])
alpha = 0.1
n_samples, n_features = X.shape
weights = np.zeros(n_features)
bias = 0.0
```

```python
y_hat = X @ weights + bias
delta_w = (1/n_samples) * X.T @ (y_hat - y) + (alpha * np.sign(weights))
delta_b = (1/n_samples) * sum(y_hat - y)
weights -= delta_w
bias -= delta_b
```

```python
def l1_regularization_gradient_descent(X: np.ndarray, y: np.ndarray, alpha: float = 0.1, learning_rate: float = 0.01, max_iter: int = 1000, tol: float = 1e-4) -> tuple:
    """
    Implement Lasso Regression using Gradient Descent.
    
    Args:
        X: Feature matrix of shape (n_samples, n_features)
        y: Target vector of shape (n_samples,)
        alpha: L1 regularization strength (higher = more regularization)
        learning_rate: Step size for gradient descent updates
        max_iter: Maximum number of iterations
        tol: Convergence tolerance - stops when L1 norm of weight gradient < tol
    
    Returns:
        tuple: (weights, bias) where:
            - weights: np.ndarray of shape (n_features,)
            - bias: float
    
    Note:
        - The bias term is NOT regularized
        - Use subgradient for L1: sign(0) = 0
    """
    n_samples, n_features = X.shape
    weights = np.zeros(n_features)
    bias = 0.0
    for i in range(max_iter):
        y_hat = X @ weights + bias
        delta_w = (1/n_samples) * X.T @ (y_hat - y) + (alpha * np.sign(weights))
        delta_b = (1/n_samples) * sum(y_hat - y)
        weights -= learning_rate * delta_w
        bias -= learning_rate * delta_b
    return weights, bias
```

```python
weights, bias = l1_regularization_gradient_descent(X, y, alpha=alpha, learning_rate=0.01, max_iter=1000)
```
